```python
import pandas as pd 
```

```python
file_path = "data/data.csv"
df = pd.read_csv(file_path, index_col= "id")
df.info()
```

```python
featuers = df.columns.to_list()[0: -1]
df[featuers].values[0]
```

```python
average_bmi = round(df["bmi"].mean(), 1)

round(average_bmi, 1)

df["bmi"] = df["bmi"].fillna(average_bmi)

        # String yazıları label'la 
df["smoking_status"] = df["smoking_status"].replace(["never smoked" , "Unknown", "formerly smoked", "smokes"], [0, 1, 2, 3])

        # Gender'ları label'la 
df["gender"] = df["gender"].replace(["Male", "Female"], [0, 1])

        # Evlenip Evlenmediklerini Label'la 
df["ever_married"] = df["ever_married"].replace(["No", "Yes"], (0, 1))

        # İş tiplerini label'la 
df["work_type"] = df["work_type"].replace(["Private", "Self-employed", "children", "Govt_job", "Never_worked"])

        # Residence_Type'ı Label'la 
df["Residence_type"] = df["Residence_type"].replace(["Urban", "Rural"], [0, 1])
```
